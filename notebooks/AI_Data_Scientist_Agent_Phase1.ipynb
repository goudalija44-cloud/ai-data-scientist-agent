# Build data_profiler.py

```python
from pathlib import Path

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/AI_Data_Scientist_Agent"
)

SRC_DIR = PROJECT_ROOT  / "src"
SRC_DIR.mkdir(parents=True, exist_ok=True)

profiler_code = '''
import pandas as pd


def profile_dataset(df: pd.DataFrame) -> dict:
    """
    Generate a reliable summary of a Pandas DataFrame.

    Returns dataset dimensions, column types, missing values,
    duplicate-row count, and numerical summaries.
    """
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    if df.empty:
        raise ValueError("Dataset must contain at least one row.")

    missing_counts = df.isna().sum()

    profile = {
        "rows": int(df.shape[0]),
        "columns": int(df.shape[1]),
        "column_names": df.columns.tolist(),
        "data_types": {
            str(column): str(dtype)
            for column, dtype in df.dtypes.items()
        },
        "missing_values": {
            str(column): int(count)
            for column, count in missing_counts.items()
        },
        "total_missing_values": int(missing_counts.sum()),
        "duplicate_rows": int(df.duplicated().sum()),
        "numerical_summary": {},
        "categorical_summary": {},
    }

    numerical_df = df.select_dtypes(include="number")
    categorical_df = df.select_dtypes(
        include=["object", "category", "bool", "string"]
    )

    if not numerical_df.empty:
        profile["numerical_summary"] = (
            numerical_df.describe().round(2).to_dict()
        )

    for column in categorical_df.columns:
        profile["categorical_summary"][str(column)] = {
            "unique_values": int(df[column].nunique(dropna=True)),
            "most_frequent": (
                str(df[column].mode(dropna=True).iloc[0])
                if not df[column].mode(dropna=True).empty
                else None
            ),
        }

    return profile
'''

profiler_path = SRC_DIR / "data_profiler.py"
profiler_path.write_text(profiler_code, encoding="utf-8")

print("Created:", profiler_path)
print("File Exist:", profiler_path.exists())
```

# Create src/data_quality.py

```python
from pathlib import Path

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/AI_Data_Scientist_Agent"
)
SRC_DIR = PROJECT_ROOT / "src"
SRC_DIR.mkdir(parents=True, exist_ok=True)

quality_code = '''
import pandas as pd


def check_data_quality(df: pd.DataFrame) -> dict:
    """
    Check common data-quality issues without modifying the dataset.
    """
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    if df.empty:
        raise ValueError("Dataset must contain at least one row.")

    missing = df.isna().sum()
    missing_columns = {
        str(column): int(count)
        for column, count in missing.items()
        if count > 0
    }

    duplicate_count = int(df.duplicated().sum())

    constant_columns = [
        str(column)
        for column in df.columns
        if df[column].nunique(dropna=False) <= 1
    ]

    potential_id_columns = [
        str(column)
        for column in df.columns
        if (
            df[column].notna().all()
            and df[column].is_unique
            and (
                "id" in str(column).lower()
                or str(column).lower().endswith("_key")
            )
        )
    ]

    findings = []

    if missing_columns:
        findings.append({
            "issue": "missing_values",
            "severity": "warning",
            "details": missing_columns,
        })

    if duplicate_count > 0:
        findings.append({
            "issue": "duplicate_rows",
            "severity": "warning",
            "details": {"count": duplicate_count},
        })

    if constant_columns:
        findings.append({
            "issue": "constant_columns",
            "severity": "info",
            "details": {"columns": constant_columns},
        })

    if potential_id_columns:
        findings.append({
            "issue": "potential_identifier_columns",
            "severity": "info",
            "details": {"columns": potential_id_columns},
        })

    return {
        "rows_checked": int(len(df)),
        "columns_checked": int(len(df.columns)),
        "missing_value_columns": missing_columns,
        "duplicate_row_count": duplicate_count,
        "constant_columns": constant_columns,
        "potential_identifier_columns": potential_id_columns,
        "findings": findings,
    }
'''

quality_path = SRC_DIR / "data_quality.py"
quality_path.write_text(quality_code, encoding="utf-8")

print("Created:", quality_path)
print("File exists:", quality_path.exists())
```

# Build the Numerical Analysis Engine

```python
# Create src/analysis_engine.py

from pathlib import Path

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/AI_Data_Scientist_Agent"
)
SRC_DIR = PROJECT_ROOT / "src"
SRC_DIR.mkdir(parents=True, exist_ok=True)

analysis_code = '''
import pandas as pd


def analyze_dataset(df: pd.DataFrame) -> dict:
    """Calculate verified descriptive statistics and correlations."""

    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    if df.empty:
        raise ValueError("Dataset must contain at least one row.")

    numeric_df = df.select_dtypes(include="number")
    results = {
        "rows": int(df.shape[0]),
        "columns": int(df.shape[1]),
        "numeric_columns": numeric_df.columns.tolist(),
        "numeric_statistics": {},
        "categorical_statistics": {},
        "correlations": {},
    }

    # Calculate numeric statistics while excluding missing values.
    for column in numeric_df.columns:
        series = numeric_df[column]
        results["numeric_statistics"][str(column)] = {
            "count": int(series.count()),
            "mean": (
                round(float(series.mean()), 4)
                if series.count() else None
            ),
            "median": (
                round(float(series.median()), 4)
                if series.count() else None
            ),
            "min": (
                float(series.min())
                if series.count() else None
            ),
            "max": (
                float(series.max())
                if series.count() else None
            ),
            "standard_deviation": (
                round(float(series.std()), 4)
                if series.count() > 1 else None
            ),
        }

    # Summarize categorical columns.
    categorical_df = df.select_dtypes(
        include=["object", "category", "bool", "string"]
    )

    for column in categorical_df.columns:
        counts = df[column].value_counts(dropna=True)
        results["categorical_statistics"][str(column)] = {
            "unique_values": int(df[column].nunique(dropna=True)),
            "top_value": str(counts.index[0]) if not counts.empty else None,
            "top_value_count": int(counts.iloc[0]) if not counts.empty else 0,
        }

    # Pearson correlations are only calculated between numeric columns.
    if len(numeric_df.columns) >= 2:
        corr = numeric_df.corr(method="pearson")
        results["correlations"] = {
            str(col): {
                str(other): (
                    round(float(corr.loc[col, other]), 4)
                    if pd.notna(corr.loc[col, other]) else None
                )
                for other in corr.columns
            }
            for col in corr.columns
        }

    return results
'''

analysis_path = SRC_DIR / "analysis_engine.py"
analysis_path.write_text(analysis_code, encoding="utf-8")

print("Created:", analysis_path)
print("File exists:", analysis_path.exists())
```

# Build the Insights Engine

Create src/insights_engine.py

```python
from pathlib import Path

insights_code = '''
import pandas as pd


def generate_insights(df: pd.DataFrame) -> dict:
    """
    Generate simple, evidence-backed descriptive insights.
    Does not infer causation or make predictions.
    """
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    if df.empty:
        raise ValueError("Dataset must contain at least one row.")

    insights = []

    # Dataset size
    insights.append({
        "type": "dataset_overview",
        "message": f"Dataset contains {len(df)} rows and {len(df.columns)} columns.",
        "evidence": {
            "rows": int(len(df)),
            "columns": int(len(df.columns)),
        },
    })

    # Missing values
    missing = df.isna().sum()
    for column, count in missing.items():
        if count > 0:
            insights.append({
                "type": "missing_values",
                "message": f"Column '{column}' contains {int(count)} missing values.",
                "evidence": {
                    "column": str(column),
                    "missing_count": int(count),
                    "missing_percentage": round(
                        float(count / len(df) * 100), 2
                    ),
                },
            })

    # Numerical patterns
    numeric_df = df.select_dtypes(include="number")

    for column in numeric_df.columns:
        series = numeric_df[column].dropna()

        if series.empty:
            continue

        if series.nunique() == 1:
            insights.append({
                "type": "constant_numeric_column",
                "message": f"Column '{column}' has only one observed numeric value.",
                "evidence": {
                    "column": str(column),
                    "value": float(series.iloc[0]),
                },
            })
            continue

        mean_value = float(series.mean())
        median_value = float(series.median())

        if mean_value > median_value:
            pattern = "mean_above_median"
        elif mean_value < median_value:
            pattern = "mean_below_median"
        else:
            pattern = "mean_equals_median"

        insights.append({
            "type": "numeric_distribution",
            "message": (
                f"Column '{column}': mean is {mean_value:.2f} "
                f"and median is {median_value:.2f}."
            ),
            "evidence": {
                "column": str(column),
                "count": int(series.count()),
                "mean": round(mean_value, 4),
                "median": round(median_value, 4),
                "pattern": pattern,
                "min": float(series.min()),
                "max": float(series.max()),
            },
        })

    # Duplicate rows
    duplicate_count = int(df.duplicated().sum())
    if duplicate_count:
        insights.append({
            "type": "duplicate_rows",
            "message": f"Dataset contains {duplicate_count} duplicate rows.",
            "evidence": {
                "duplicate_count": duplicate_count,
            },
        })

    return {
        "insight_count": len(insights),
        "insights": insights,
    }
'''

insights_path = PROJECT_ROOT / "src" / "insights_engine.py"
insights_path.write_text(insights_code, encoding="utf-8")

print("Created:", insights_path)
print("File exists:", insights_path.exists())
```

# Build a safe CSV data loader

Create src/data_loader.py

```python
from pathlib import Path

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/AI_Data_Scientist_Agent"
)

loader_code = '''
from pathlib import Path
import pandas as pd

MAX_FILE_SIZE_BYTES = 50 * 1024 * 1024  # 50 MB

def load_csv(file_path: str | Path) -> pd.DataFrame:
  """Load and validate a CSV file without modifying its contents."""

  path = Path(file_path)

  if not path.exists():
    raise FileNotFoundError(f"File Not Found: {file_path}")

  if not path.is_file():
    raise ValueError("The provided path is not a file.")

  if path.suffix.lower() != '.csv':
    raise ValueError("Only CSV files are supported.")

  if path.stat().st_size > MAX_FILE_SIZE_BYTES:
    raise ValueError("File size exceeds the limit of 50 MB.")

  if path.stat().st_size == 0:
    raise ValueError("File is empty.")


  try:

    df = pd.read_csv(file_path)
  except pd.errors.EmptyDataError as exc:
    raise ValueError("CSV contains no tabular data.") from exc

  except (pd.errors.ParserError, UnicodeDecodeError) as exc:
    raise ValueError("CSV file could not be parsed.") from exc


  if df.empty:
    raise ValueError("CSV contains headers but no data rows.")

  if len(df.columns) == 0:
    raise ValueError("CSV contains no columns.")

  if df.columns.duplicated().any():
    raise ValueError("CSV contains duplicate column names.")

  if any(not str(col).strip() for columns in df.columns):
    raise ValueError("CSV contains an empty column name.")

  return df

'''

loader_path = PROJECT_ROOT / "src"  / "data_loader.py"
loader_path.write_text(loader_code, encoding="utf-8")

print("Created:", loader_path)
print("file exists:", loader_path.exists())
```

```python
from pathlib import Path

TESTS_DIR = PROJECT_ROOT / "tests"
TESTS_DIR.mkdir(parents=True, exist_ok=True)

loader_tests = '''
import pandas as pd
import pytest

from data_loader import load_csv
from run_dataset_analysis import analyze_csv_file


def test_load_valid_csv(tmp_path):
    path = tmp_path / "valid.csv"
    pd.DataFrame({
        "Age": [20, 30],
        "Salary": [25000, 40000],
    }).to_csv(path, index=False)

    df = load_csv(path)

    assert df.shape == (2, 2)
    assert df["Age"].tolist() == [20, 30]


def test_missing_file_is_rejected(tmp_path):
    with pytest.raises(FileNotFoundError):
        load_csv(tmp_path / "missing.csv")


def test_non_csv_file_is_rejected(tmp_path):
    path = tmp_path / "data.txt"
    path.write_text("example", encoding="utf-8")

    with pytest.raises(ValueError, match="Only CSV"):
        load_csv(path)


def test_empty_file_is_rejected(tmp_path):
    path = tmp_path / "empty.csv"
    path.write_text("", encoding="utf-8")

    with pytest.raises(ValueError):
        load_csv(path)


def test_headers_without_rows_are_rejected(tmp_path):
    path = tmp_path / "headers_only.csv"
    path.write_text("Age,Salary\\\\n", encoding="utf-8")

    with pytest.raises(ValueError):
        load_csv(path)


def test_end_to_end_analysis(tmp_path):
    path = tmp_path / "customers.csv"
    pd.DataFrame({
        "Age": [20, 30, 30],
        "City": ["Thane", "Mumbai", "Mumbai"],
    }).to_csv(path, index=False)

    report = analyze_csv_file(path)

    assert report["status"] == "success"
    assert report["source_file"] == "customers.csv"
    assert report["dataset_profile"]["rows"] == 3
    assert report["data_quality"]["duplicate_row_count"] == 1
'''

test_path = TESTS_DIR / "test_data_loader.py"
test_path.write_text(loader_tests, encoding="utf-8")

print("Created:", test_path)
print("File exists:", test_path.exists())
```
